**Sadə List Tapşırıqları**

fruits= ["alma", "armud","banan", "gilas"]-list-i yaradın və çap edin.

```python
fruits = ["alma", "armud", "banan", "gilas"]
print(fruits)
```

Bu list-ə "albuxara" sözünü əlavə edin ( append ).

```python
fruits = ["alma", "armud", "banan", "gilas"]
fruits.append("albuxara")
print(fruits)
```

List-in ilk elementini çap edin.

```python
fruits = ["alma", "armud", "banan", "gilas", "albuxara"]
print(fruits[0])
```

List-in son elementini çap edin.

```python
fruits = ["alma", "armud", "banan", "gilas", "albuxara"]
print(fruits[-1])
```

List-in uzunluğunu ( len) tapın.

```python
fruits = ["alma", "armud", "banan", "gilas", "albuxara"]
print(len(fruits))
```

List-i tərsinə çevirin (reverse()

```python
fruits = ["alma", "armud", "banan", "gilas", "albuxara"]
fruits.reverse()
print(fruits)
```

 List-i əlifba sırası ilə düzün (sortO).

```python
fruits = ["alma", "armud", "banan", "gilas", "albuxara"]
fruits.sort()
print(fruits)
```

List-dən "banan" sözünü silin (remove()).

```python
fruits = ["alma", "armud", "banan", "gilas", "albuxara"]
fruits.remove("banan")
print(fruits)
```

İki list-i birləşdirin: list1 + list2.

```python
fruits = ["alma", "armud", "banan", "gilas", "albuxara"]
fruits2 = ["heyva", "nar"]
birlesmis_list = fruits + fruits2
print(birlesmis_list)
```

List-i2-cielementdən 4-cü elementə qədər dilimləyin (slicing ).

```python

fruits = ['alma', 'albuxara', 'armud', 'gilas', 'heyva', 'nar']
dilimlenmis_list = fruits[1:4]
print(dilimlenmis_list)
```

**Sadə Tuple Tapşırıqları **

colors = ("qırmızı","yaşıl", "mavi")-tuple yaradın və çap edin.

```python
colors = ("qırmızı", "yaşıl", "mavi")
print(colors)
```

Tuple-in ilk elementini çap edin.

```python
colors = ("qırmızı", "yaşıl", "mavi")
print(colors[0])
```

Tuple-in uzunluğunu tapın

```python
colors = ("qırmızı", "yaşıl", "mavi")
print(len(colors))
```

Tuple-1 list-ə çevirin ( list(colors))

```python
colors_list = list(colors)
print(colors_list)
print(type(colors_list))
```

List-i geriyə tuple-a çevirin ( tuple(list)).

```python
colors_tuple = tuple(colors_list)
print(colors_tuple)
print(type(colors_tuple))
```

İki tuple-1 birləşdirin: tuplel + tuple2.

```python
colors = ("qırmızı", "yaşıl", "mavi")
tuple2 = ("sarı", "qara")
birlesmis_tuple = colors + tuple2
print(birlesmis_tuple)
```

Tuple-1 dəyişənlərə ayırın (unpacking): a, b, c= colors.

```python
colors = ("qırmızı", "yaşıl", "mavi")
a, b, c = colors
print("a:", a)
print("b:", b)
print("c:", c)
```

Tuple-in içində "mavi" olub-olmadığın1 yoxlamadan sadəcə index()ilə yerini tapın.

```python
colors = ("qırmızı", "yaşıl", "mavi")
index_mavi = colors.index("mavi")
print(index_mavi)
```
