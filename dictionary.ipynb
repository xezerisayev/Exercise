bir dictionary-de telebenin adi ve qiymeti (0-100) saxlanilir:
student = {"ad": "Elvin", "qiymet": 75}
eger qiymet 90 dan yuxaridirsa "Ela" 70-90 arasi"Yaxsi",50-70 arasi"Kafi",asagi ise "Ikiletdi" yazdirin

```python
student = {"ad": "Elvin", "qiymet": 75}

qiymet = student["qiymet"]

if qiymet > 90:
    print("Əla")
elif 70 <= qiymet <= 90:
    print("Yaxşı")
elif 50 <= qiymet < 70:
    print("Kafi")
else:
    print("İkiletdi")
```

Dictionary-de saxlanilan yasa gore Şexsin"usaq"(0-12),"yeniyetme"(13-17),"yasli"(18-64),"ahil"(65+)oldugunu mueyyen edin:
person={"ad":"Aygun","yas":34}

```python
person = {"ad": "Aygun", "yas": 34}

yas = person["yas"]

if 0 <= yas <= 12:
    qrup = "uşaq"
elif 13 <= yas <= 17:
    qrup = "yeniyetmə"
elif 18 <= yas <= 64:
    qrup = "yaşlı"
elif yas >= 65:
    qrup = "ahıl"
else:
    qrup = "Düzgün yaş daxil edilməyib"

print(f"{person['ad']} - {qrup} ({yas} yaş)")
```

bank hesablari dictionaryde saxlanilir:
hesablar = {"Ali":500,"Vali":-100,"Aygun":0}
her hesab ucun:menfi balansdirsa"Borclu",0-dirsa"Bos hesab",musbetdirse"Aktiv"yazdirin

```python
hesablar = {"Ali": 500, "Vali": -100, "Aygun": 0}

for ad, balans in hesablar.items():
    if balans < 0:
        status = "Borclu"
    elif balans == 0:
        status = "Boş hesab"
    else:  # balans > 0
        status = "Aktiv"

    print(f"{ad}: {status} (Balans: {balans})")
```

istifadeciden dictionary-ye acar-deyer elave etmek istenilir.eger hemin acar artiq ductionary-de varsa,"Bu acar movcuddur,uzerine yazilsin?"-deye sual verin,yoxdursa birbasa elave edin

```python

my_dict = {"ad": "Aygun", "yas": 34, "balans": 500}
print("Mövcud lüğət:", my_dict)
yeni_acar = input("Əlavə etmək istədiyiniz açarı daxil edin: ")
yeni_deyer = input("Dəyəri daxil edin: ")
if yeni_acar in my_dict:
    cavab = input(f"'{yeni_acar}' açarı artıq mövcuddur, üzərinə yazılşın? (hə/yox): ").strip().lower()
    if cavab in ("hə", "he", "yes", "y"):
        my_dict[yeni_acar] = yeni_deyer
        print("Açarın dəyəri yeniləndi.")
    else:
        print("Əməliyyat ləğv edildi, açar dəyişdirilmədi.")
else:
    my_dict[yeni_acar] = yeni_deyer
    print("Yeni açar və dəyər uğurla əlavə edildi.")
print("Lüğətin son vəziyyəti:", my_dict)
```

hava haqqinda melumat dictionary-de saxlanilir:
hava = {"temperator": 32, "yagish":False}
eger temperator 30-dan yuxaridirsa ve yagis yoxdursa-"Isti ve gunesli",yagis varsa-"Isti amma yagisli",temperator asagidisa-"serin hava"yazdirin

```python
hava = {"temperator": 32, "yagish": False}

temp = hava["temperator"]
yagish = hava["yagish"]

if temp > 30:
    if not yagish:  # yagish == False
        print("İsti və günəşli")
    else:  # yagish == True
        print("İsti amma yağışlı")
else:  # temp <= 30
    print("Sərin hava")
```

Bir neçə işçinin maaşı dictionary-də saxlanılır:
python
maaslar ={"Ali":1500,"Vali": 2500, "Aygun":1800,"Nigar": 3200}
Maaşı 2000-dən yuxarı olanları ayrıca yeni dictionary-yə yığın
Bu işçilərə görə (maaşdan asılı olaraq) mükafat faizini müəyyən edin:3000-dən yuxarı-→ 15%, 2000-3000→10%, aşağı→5%
· Yeni maaşları hesablayıb dictionary-də yeniləyin

```python
maaslar = {"Ali": 1500, "Vali": 2500, "Aygun": 1800, "Nigar": 3200}
yuksek_maaslilar = {ad: maas for ad, maas in maaslar.items() if maas > 2000}
print("Maaşı 2000-dən yuxarı olanlar:", yuksek_maaslilar)
yeni_maaslar = {}
for ad, maas in yuksek_maaslilar.items():
  if maas > 3000:
    faiz = 0.15
  elif maas >= 2000:
    faiz = 0.10
  else:
    faiz = 0.05
  yeni_maaslar[ad] = int(maas + (maas * faiz))
print("Mükafatlandırılmış yeni maaşlar:", yeni_maaslar)
```

Login sistemi: istifadəçi adları və şifrələri dictionary-də saxlanılır:
python
users = {"admin":"1234", "ali":"qwerty"}
İstifadəçidən ad və şifrə soruşulur:
·Ad dictionary-də yoxdursa → "İstifadəçi tapılmadı" · Ad var, ifrə yanlışdırsa→"Şifrə səhvdir" · Hər ikisi doğrudursa →"Giriş uğurludur"

```python
users = {"admin": "1234", "ali": "qwerty"}
daxil_edilen_ad = input("İstifadəçi adını daxil edin: ")
daxil_edilen_sifre = input("Şifrəni daxil edin: ")
if daxil_edilen_ad not in users:
  print("İstifadəçi tapılmadı")
elif users[daxil_edilen_ad] != daxil_edilen_sifre:
  print("Şifrə səhvdir")
else:
  print("Giriş uğurludur")
```

Anbar idarəetmə proqramı: məhsullar və sayları dictionary-də saxlanılır:
python
anbar= {"alma": 20,"armud": 5,"banan": 0 }
İstifadəçi məhsul adı və almaq istədiyi miqdarı daxil edir:
·Məhsul yoxdursa→"Belə məhsul yoxdur" Stok kifayət deyilsə→ "Kifayət qədər stok yoxdur"
1 Kifayətdirsə→ miqdarı stokdan çıxarın və "Alış uğurla tamamlandı" yazın

```python
anbar = {"alma": 20, "armud": 5, "banan": 0}
mehsul = input("Almaq istədiyiniz məhsulun adı: ").strip().lower()
miqdar = int(input("Almaq istədiyiniz miqdar: "))
if mehsul not in anbar:
  print("Belə məhsul yoxdur")
elif anbar[mehsul] < miqdar:
  print("Kifayət qədər stok yoxdur")
else:
  anbar[mehsul] -= miqdar
  print("Alış uğurla tamamlandı")
  print(f"'{mehsul}' üzrə qalan stok: {anbar[mehsul]}")
```
