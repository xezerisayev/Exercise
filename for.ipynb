1-den 50 ye qeder ededleri yoxlayin.Cut ededlerin cemini ve tek ededlerin sayini ayri ayriliqda tapib cap edin.

```python
cut_cem = 0
tek_say = 0

for i in range(1, 51):  # 1-dən 50 daxil olmaqla
    if i % 2 == 0:
        cut_cem += i
    else:
        tek_say += 1

print("Cüt ədədlərin cəmi:", cut_cem)
print("Tək ədədlərin sayı:", tek_say)
```

istifadeciden n ededi alin ve for dovru ile n! (faktorial) hesablayin
numune:5!=120

```python
n = int(input("Bir ədəd daxil edin: "))
faktorial = 1

for i in range(1, n + 1):
    faktorial *= i

print(f"{n}! = {faktorial}")
```

asagidaki siyahidav max() funkisyasindan istifade etmeden yanliz for dovru ile en boyuk ve en kicik ededleri tapin

```python
ededler = [14, 3, 78, 25, 9, 56, 41]

en_boyuk = ededler[0]
en_kicik = ededler[0]

for eded in ededler:
    if eded > en_boyuk:
        en_boyuk = eded
    if eded < en_kicik:
        en_kicik = eded

print("Ən böyük ədəd:", en_boyuk)
print("Ən kiçik ədəd:", en_kicik)
```

istifadeciden bir soz daxil etmesini isteyin.for dovru ile hemin sozu tersine yazin (slice [::-1] istifade etmeyin).sonra sozun palindron olub-olmadigini yoxlayin.
numune:ATA->ATA(palindromdur)

```python
soz = input("Bir söz daxil edin: ")
ters_soz = ""

for herf in soz:
    ters_soz = herf + ters_soz

print("Tərs söz:", ters_soz)

if soz.lower() == ters_soz.lower():
    print(f"{soz} -> {ters_soz} (palindromdur)")
else:
    print(f"{soz} -> {ters_soz} (palindrom deyil)")
```

istifadeciden n ededi alin ve Fibonacci ardicilliginin ilk n heddinin ardicilligini cap edin
Numune:(n=8 ucun) 0,1,1,2,3,5,8,13

```python
n = int(input("Neçə Fibonaççi həddi çap olunsun? "))

a, b = 0, 1
fibonacci = []

for _ in range(n):
    fibonacci.append(a)
    a, b = b, a + b

print(f"n = {n} üçün:", fibonacci)
```
