Tələbə haqqında dictionary yarat və onun adını çap et.

```python
telebe = {
    "ad": "Elvin",
    "soyad": "Məmmədov",
    "yas": 21,
    "ixtisas": "Kompüter elmləri"
}
print(telebe["ad"])
```

 Dictionary-yə yeni açar "seher"və dəyər "Bakı" əlavə et.

```python
telebe = {
    "ad": "Elvin",
    "soyad": "Məmmədov",
    "yas": 21,
    "ixtisas": "Kompüter elmləri"
}
telebe["seher"] = "Bakı"
print(telebe)
```

Dictionary-dəki bütün açar və dəyərləri çap et.

```python
telebe = {
    "ad": "Elvin",
    "soyad": "Məmmədov",
    "yas": 21,
    "ixtisas": "Kompüter elmləri",
    "seher": "Bakı"
}
print("Bütün açar və dəyərlər:")
print(telebe.items())
print("\nSəliqəli siyahı:")
for acar, deyer in telebe.items():
    print(f"{acar}: {deyer}")
```

 Dictionary-də "il" açarı varsa "Məlumat var", yoxdursa "Yoxdur" yaz.

```python
telebe = {
    "ad": "Elvin",
    "soyad": "Məmmədov",
    "yas": 21,
    "ixtisas": "Kompüter elmləri",
    "seher": "Bakı"
}
if "il" in telebe:
    print("Məlumat var")
else:
    print("Yoxdur")
```
